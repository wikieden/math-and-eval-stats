## 定义

```python
import math
import numpy as np
```

`expectation`：第一步：期望 E[X] = Σ x·P(x)。

```python
def expectation(values, probs) -> float:
    """第一步：期望 E[X] = Σ x·P(x)。"""
    return float(sum(v * p for v, p in zip(values, probs)))
```

`variance`：第二步：方差 Var(X) = E[(X - E[X])²] = E[X²] - E[X]²。

```python
def variance(values, probs) -> float:
    """第二步：方差 Var(X) = E[(X - E[X])²] = E[X²] - E[X]²。"""
    m = expectation(values, probs)
    return float(sum(p * (v - m) ** 2 for v, p in zip(values, probs)))
```

`bayes_posterior`：第三步：P(真成功 | 检测器说成功) = P(说成功 | 真成功)·P(真成功) / P(说成功)。

```python
def bayes_posterior(prior: float, sensitivity: float, specificity: float) -> float:
    """第三步：P(真成功 | 检测器说成功) = P(说成功 | 真成功)·P(真成功) / P(说成功)。"""
    says_yes = sensitivity * prior + (1 - specificity) * (1 - prior)
    return sensitivity * prior / says_yes
```

`bernoulli_log_likelihood`：第四步：伯努利对数似然 log L(p) = k·log p + (n-k)·log(1-p)。

```python
def bernoulli_log_likelihood(p: float, k: int, n: int) -> float:
    """第四步：伯努利对数似然 log L(p) = k·log p + (n-k)·log(1-p)。"""
    return k * math.log(p) + (n - k) * math.log(1 - p)
```

`mle_bernoulli`：第四步：令 d/dp log L = k/p - (n-k)/(1-p) = 0，得 p̂ = k/n。

```python
def mle_bernoulli(k: int, n: int) -> float:
    """第四步：令 d/dp log L = k/p - (n-k)/(1-p) = 0，得 p̂ = k/n。"""
    return k / n
```

`map_bernoulli`：第五步：Beta(a, b) 先验下的 MAP = (k + a - 1) / (n + a + b - 2)。

```python
def map_bernoulli(k: int, n: int, a: float, b: float) -> float:
    """第五步：Beta(a, b) 先验下的 MAP = (k + a - 1) / (n + a + b - 2)。"""
    return (k + a - 1) / (n + a + b - 2)
```

`bc_nll`：第六步：行为克隆的损失 = 示范动作的平均负对数概率（离散动作）。

```python
def bc_nll(probs_of_demo_actions) -> float:
    """第六步：行为克隆的损失 = 示范动作的平均负对数概率（离散动作）。"""
    return float(-np.mean(np.log(probs_of_demo_actions)))
```

## 一步一步跑

### 第一、二步 期望与方差

```python
expectation([1, -1, 0], [0.7, 0.1, 0.2]), variance([1, -1, 0], [0.7, 0.1, 0.2])
```

### 第三步 贝叶斯

```python
bayes_posterior(0.7, 0.95, 0.90), bayes_posterior(0.2, 0.95, 0.90)
```

### 第四步 对数似然与 MLE

```python
[(p, round(bernoulli_log_likelihood(p, 21, 30), 3)) for p in (0.5, 0.6, 0.7, 0.8)], mle_bernoulli(21, 30)
```

### 第五步 MAP

```python
map_bernoulli(21, 30, 2, 2), map_bernoulli(3, 3, 2, 2)
```

### 第六步 行为克隆的 NLL

```python
bc_nll([0.9, 0.6, 0.3])
```

## 代码里才看得到的

1. p = 0 或 1 时 log(0) 报错；真实代码用 log_softmax / logsigmoid。
2. 对数似然随 n 线性变大，loss 通常除以 N。
3. MAP 公式只适用于 a, b ≥ 1。
